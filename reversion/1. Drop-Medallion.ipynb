```sql
%python
dbutils.widgets.removeAll()
```

```sql
%python
dbutils.widgets.text("srcData1", "adlsrcset1")
dbutils.widgets.text("catalogo", "catalog_au")
```

```sql
%python

srcData1 = dbutils.widgets.get("srcData1")
catalogo = dbutils.widgets.get("catalogo")


ruta_bronze = f"abfss://bronze@{srcData1}.dfs.core.windows.net"
ruta_silver = f"abfss://silver@{srcData1}.dfs.core.windows.net"
ruta_golden = f"abfss://golden@{srcData1}.dfs.core.windows.net"
```

### Eliminacion tablas Bronze

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.bronze.clientes;
DROP TABLE IF EXISTS ${catalogo}.bronze.transacciones;
```

```sql
%python
dbutils.fs.rm(f"{ruta_bronze}/clientes", True)
dbutils.fs.rm(f"{ruta_bronze}/transacciones", True)
```

### Eliminacion tablas Silver

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.silver.transacciones_clientes_silver;
DROP TABLE IF EXISTS ${catalogo}.silver.clientes_cleaned;
DROP TABLE IF EXISTS ${catalogo}.silver.transacciones_cleaned;
```

```sql
%python
# REMOVE DATA (Silver)
dbutils.fs.rm(f"{ruta_silver}/transacciones_clientes_silver", True)
dbutils.fs.rm(f"{ruta_silver}/clientes_cleaned", True)
dbutils.fs.rm(f"{ruta_silver}/transacciones_cleaned", True)
```

### Eliminacion tablas Golden

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.golden.golden_analisis_fraude;
```

```sql
%python
# REMOVE DATA (Golden)
dbutils.fs.rm(f"{ruta_golden}/golden_analisis_fraude", True)
```

```sql
%python
print(f"✅ Se borraron exitosamente las tablas y archivos de {catalogo} en los contenedores bronze, silver y golden.")
```
